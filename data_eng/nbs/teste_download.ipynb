```python
from SpectraHitran.SpectraGenerator.spectragenerator import SpectraGenerator

from hapi import SLIT_TRIANGULAR, SLIT_MICHELSON
```

```python
generator = SpectraGenerator()
```

```python
generator.downloadMolecule(diluents=('self',),
                           moleculeName='hcl',
                           isotopologues=(52,53),
                           wavenumberRange=(4_000,50_000))
```

```python
generator.simulateSpectra(moleculeName='hcl',
                          diluent={'self': 1},
                          enviroment={'p':0.2,'T':300 ,'l': 5},
                          step=0.01,
                          convolve=False,
                          resolution=1,
                          slit_function=SLIT_TRIANGULAR)
```

```python
_ = generator.plot(interactive=True)
```
